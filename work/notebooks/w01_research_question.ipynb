## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

This work can describe observed and measured relationships between search and traffic signals. The results can provide directional insights and decision-support for prioritizing content for review.

This work cannot prove that one signal causes another outcome. It cannot provide causal proof, guarantee future performance, or claim to predict Google's search results. Any model developed later will be treated as a decision-support tool rather than a guarantee of content performance.
